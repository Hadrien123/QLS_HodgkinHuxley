```python
# %% SECTION 1: Packages import
import os
import numpy as np

# Import other packages as necessary
import matplotlib.pyplot as plt
from sklearn.model_selection import GridSearchCV
from scipy.optimize import least_squares
```

```python
# %% SECTION 2: Helper functions
# Whole-trace parameters: [e_r, g_bar, tau_m nodes, tau_h nodes, m_inf nodes, h_inf nodes]
rng = np.random.default_rng(0)
n_starts = 20

def gate_model(tau, x_0, x_inf, dt):
    return x_0 + (x_inf - x_0) * (1 - np.exp(-dt / tau))

def get_current_segment(e_r, g_bar, tau_m, tau_h, x_inf, v, t, m_0, h_0):
    m = np.zeros(len(v))
    h = np.zeros(len(v))
    m[0] = m_0
    h[0] = h_0
    for i in range(1, len(t)):
        dt = t[i] - t[i-1]
        m[i] = gate_model(tau_m, m[i-1], x_inf, dt)
        h[i] = gate_model(tau_h, h[i-1], x_inf, dt)
    return g_bar * (v - e_r) * m * h, m[-1], h[-1]

def compute_error_segment(params, t, v, m_0, h_0, test_i):
    e_r, g_bar, tau_m, tau_h, x_inf = params
    current, _, _ = get_current_segment(e_r, g_bar, tau_m, tau_h, x_inf, v, t, m_0, h_0)
    return current - test_i

# Interpolate both time constants and steady-state gate values over voltage.
def get_current(e_r, g_bar, tau_m, tau_h, m_inf, h_inf, v, t, v_steps):
    m = np.zeros(len(v))
    h = np.zeros(len(v))
    tau_m_step = np.interp(v, v_steps, tau_m)
    tau_h_step = np.interp(v, v_steps, tau_h)
    m_inf_step = np.interp(v, v_steps, m_inf)
    h_inf_step = np.interp(v, v_steps, h_inf)
    m[0] = 0
    h[0] = 1
    for i in range(1, len(t)):
        dt = t[i] - t[i-1]
        m[i] = gate_model(tau_m_step[i], m[i-1], m_inf_step[i], dt)
        h[i] = gate_model(tau_h_step[i], h[i-1], h_inf_step[i], dt)
    return g_bar * (v - e_r) * m * h

def compute_error(params, t, v, test_i, v_steps):
    n_steps = len(v_steps)
    e_r, g_bar = params[:2]
    tau_m = params[2:2+n_steps]
    tau_h = params[2+n_steps:2+2*n_steps]
    m_inf = params[2+2*n_steps:2+3*n_steps]
    h_inf = params[2+3*n_steps:2+4*n_steps]
    return get_current(e_r, g_bar, tau_m, tau_h, m_inf, h_inf, v, t, v_steps) - test_i

def whole_trace_bounds(n_steps):
    lower = [-np.inf, -np.inf] + [0.01] * n_steps + [0.01] * n_steps + [-np.inf] * (2*n_steps)
    upper = [np.inf, np.inf] + [1e3] * n_steps + [1e4] * n_steps + [np.inf] * (2*n_steps)
    return lower, upper

def random_start(n_steps):
    e_r = rng.uniform(-100, 50)
    g_bar = 10 ** rng.uniform(-1, 1.5)
    tau_m = 10 ** rng.uniform(-1, 1, n_steps)
    tau_h = 10 ** rng.uniform(0, 2, n_steps)
    m_inf = np.sort(rng.uniform(0, 1, n_steps))
    h_inf = np.sort(rng.uniform(0, 1, n_steps))[::-1]
    return [e_r, g_bar, *tau_m, *tau_h, *m_inf, *h_inf]

def gaussian(scale, center, width, v):
    return scale * np.exp(-((v-center)**2) / width)

def compute_error_tau(params, v, tau):
    scale, center, width = params
    return gaussian(scale, center, width, v) - tau

def sigmoid(steepness, center, v):
    return 1 / (1 + np.exp(-steepness * (v-center)))

def compute_error_x_inf(params, v, x_inf):
    steepness, center = params
    return sigmoid(steepness, center, v) - x_inf

def plot_x_inf(x_inf, v):
    fit = least_squares(compute_error_x_inf, args=(v, x_inf), x0=[1, 0])
    voltages = [i for i in range(-100, 21)]
    plt.figure()
    plt.subplot(1, 1, 1)
    plt.grid(True)
    plt.scatter(v, x_inf, linewidth=2)
    plt.plot(voltages, sigmoid(fit.x[0], fit.x[1], voltages))
    plt.ylabel("x_inf")
    plt.xlabel("Voltage (mV)")
    plt.legend()

def plot_taus(v, tau_m, tau_h):
    v = np.asarray(v)
    tau_m = np.asarray(tau_m)
    tau_h = np.asarray(tau_h)
    v_unique = np.unique(v)
    tau_m_unique = np.array([np.mean(tau_m[v == voltage]) for voltage in v_unique])
    tau_h_unique = np.array([np.mean(tau_h[v == voltage]) for voltage in v_unique])
    voltages = np.arange(-100, 21)
    plt.figure()
    plt.subplot(2, 1, 1)
    plt.grid(True)
    plt.scatter(v, tau_m, linewidth=2)
    plt.plot(voltages, np.interp(voltages, v_unique, tau_m_unique))
    plt.ylabel("tau_m (msec)")
    plt.subplot(2, 1, 2)
    plt.grid(True)
    plt.scatter(v, tau_h, linewidth=2)
    plt.plot(voltages, np.interp(voltages, v_unique, tau_h_unique))
    plt.ylabel("tau_h (msec)")
    plt.xlabel("Voltage (mV)")

def compute_aic(residuals, n_params):
    n = len(residuals)
    ss_res = np.sum(np.asarray(residuals) ** 2)
    return n * np.log(ss_res / n) + 2 * n_params

def compute_bic(residuals, n_params):
    n = len(residuals)
    ss_res = np.sum(np.asarray(residuals) ** 2)
    return n * np.log(ss_res / n) + n_params * np.log(n)

def fit_holdout(x0, t, v, test_i, v_steps, train_mask):
    starts = [x0] + [random_start(len(v_steps)) for _ in range(n_starts - 1)]
    bounds = whole_trace_bounds(len(v_steps))
    fits = [least_squares(lambda p: compute_error(p, t, v, test_i, v_steps)[train_mask], x0=s, bounds=bounds, max_nfev=40) for s in starts]
    fit = min(fits, key=lambda result: result.cost)
    res = compute_error(fit.x, t, v, test_i, v_steps)
    r2 = lambda mask: 1 - np.sum(res[mask] ** 2) / np.sum((test_i[mask] - np.mean(test_i[mask])) ** 2)
    return r2(train_mask), r2(~train_mask), np.sqrt(np.mean(res[~train_mask] ** 2))
```

```python
# %% SECTION 3: Main function
def run_vclamp_model(folder: str, data: str = "CookLab1UnknownChannelExtra.txt", run_validation: bool = False):
    """
    Compute current from voltage-clamp input.

    Parameters
    ----------
    folder :
        Directory with input file for voltage-clamp model.
    data :
        Name of input file.
    run_validation :
        If True, run the train/test hold-out experiments (slow).

    Returns
    -------
    np.ndarray
        Output current computed by the voltage-clamp model.
    """
    # %% SECTION 3-a: Load provided input data.
    # NO MODIFICATIONS: This section should be kept as is.
    data_extra = np.loadtxt(os.path.join(folder, "data/CookLab1UnknownChannelExtra.txt"))
    tau_m = []
    tau_h = []
    tau_v = []
    list_i = []
    x_inf = []
    m_0 = []
    h_0 = []
    past = 1
    segment_lower = [-np.inf, -np.inf, 0.01, 0.01, -np.inf]
    segment_upper = [np.inf, np.inf, 1e3, 1e4, np.inf]

    for j in range(1, 11):
        start = j*100-past
        stop = (j+1)*100
        t = data_extra[start:stop, 0]
        v = data_extra[start:stop, 1]
        test_i = data_extra[start:stop, 2]

        if j == 1:
            m_0.append(0)
            h_0.append(1)
        fit = least_squares(
            compute_error_segment,
            args=(t[past:], v[past:], m_0[-1], h_0[-1], test_i[past:]),
            x0=[-60, 10, 1, 1, 0.5],
            bounds=(segment_lower, segment_upper),
        )

        current, m_last, h_last = get_current_segment(
            fit.x[0], fit.x[1], fit.x[2], fit.x[3], fit.x[4], v, t, m_0[-1], h_0[-1]
        )
        m_0.append(m_last)
        h_0.append(h_last)

        print(f"Fitted parameters for V={v[0]}, V={v[-1]}:", fit.x)
        if j % 2 == 0:
            tau_v.append(tau_v[-1])
            tau_m.append(fit.x[2])
            tau_h.append(fit.x[3])
            x_inf.append(fit.x[4])
        else:
            tau_m.append(fit.x[2])
            tau_h.append(fit.x[3])
            tau_v.append(v[past])
            x_inf.append(fit.x[4])
            print("Fit succeeded:", fit.success)
        list_i += list(current)

    tau_v_values = np.asarray(tau_v)
    tau_m_values = np.asarray(tau_m)
    tau_h_values = np.asarray(tau_h)
    plot_taus(tau_v_values, tau_m_values, tau_h_values)
    plot_x_inf(x_inf, tau_v_values)

    # Average repeated segment estimates at each voltage, then interpolate them
    # onto the whole-trace model's voltage nodes.
    tau_v_unique = np.unique(tau_v_values)
    tau_m_unique = np.array([np.mean(tau_m_values[tau_v_values == voltage]) for voltage in tau_v_unique])
    tau_h_unique = np.array([np.mean(tau_h_values[tau_v_values == voltage]) for voltage in tau_v_unique])

    data = np.loadtxt(os.path.join(folder, "data/CookLab1UnknownChannelExtra.txt"))
    t = data[:, 0]
    v = data[:, 1]
    test_i = data[:, 2]

    v_steps = [-100, -60, -40, -20, 0, 20]
    initial_tau_m = np.interp(v_steps, tau_v_unique, tau_m_unique)
    initial_tau_h = np.interp(v_steps, tau_v_unique, tau_h_unique)
    initial_m_inf = [0.01, 0.05, 0.2, 0.5, 0.8, 0.95]
    initial_h_inf = [0.99, 0.95, 0.8, 0.5, 0.2, 0.05]
    x0 = [-60, 10, *initial_tau_m, *initial_tau_h, *initial_m_inf, *initial_h_inf]
    whole_bounds = whole_trace_bounds(len(v_steps))

    fits = []
    for k in range(n_starts):
        start_params = x0 if k == 0 else random_start(len(v_steps))
        fit = least_squares(
            compute_error,
            x0=start_params,
            args=(t, v, test_i, v_steps),
            bounds=whole_bounds,
            max_nfev=40,
        )
        fits.append(fit)
        print(f"start {k}: cost = {fit.cost:.4f}, success = {fit.success}, params = {fit.x}")
    fit = min(fits, key=lambda result: result.cost)
    print("Best parameters:", fit.x)
    n_steps = len(v_steps)
    i = get_current(
        fit.x[0], fit.x[1],
        fit.x[2:2+n_steps],
        fit.x[2+n_steps:2+2*n_steps],
        fit.x[2+2*n_steps:2+3*n_steps],
        fit.x[2+3*n_steps:2+4*n_steps],
        v, t, v_steps,
    )

    ss_res = np.sum((test_i - i) ** 2)
    ss_tot = np.sum((test_i - np.mean(test_i)) ** 2)
    r_squared = 1 - (ss_res / ss_tot)
    print("R-squared:", r_squared)
    print("AIC:", compute_aic(test_i - i, len(fit.x)))
    print("BIC:", compute_bic(test_i - i, len(fit.x)))
    print("Fitted parameters:", fit.x)
    print("Fit succeeded:", fit.success)

    if run_validation:
        idx = np.arange(len(t))
        for frac in [0.05, 0.10, 0.20, 0.30, 0.50]:
            train_r2, test_r2, test_rmse = fit_holdout(x0, t, v, test_i, v_steps, idx < len(t) * (1 - frac))
            print(f"tail {frac:.0%} held out: train R2 = {train_r2:.3f}, test R2 = {test_r2:.3f}, test RMSE = {test_rmse:.4f}")
        for j in [1, 3, 5, 7, 9]:
            train_r2, test_r2, test_rmse = fit_holdout(x0, t, v, test_i, v_steps, (idx < j*100) | (idx >= (j+1)*100))
            print(f"V = {v[j*100]:.0f} mV step held out: train R2 = {train_r2:.3f}, test R2 = {test_r2:.3f}, test RMSE = {test_rmse:.4f}")

    plt.figure()
    plt.subplot(2, 1, 1)
    plt.grid(True)
    plt.plot(data_extra[:1100, 0], data_extra[:1100, 1], linewidth=2)
    plt.axis("tight")
    plt.ylabel("V step (mV)")

    plt.subplot(2, 1, 2)
    plt.grid(True)
    plt.plot(data_extra[:1100, 0], data_extra[:1100, 2], label="Measured current")
    plot_t = []
    for j in range(1, 11):
        start = j*100-past
        stop = (j+1)*100
        plot_t += list(data_extra[start:stop, 0])
    plt.scatter(plot_t, list_i, label="Model current", color="r")
    plt.xlabel("Time (msec)")
    plt.ylabel("Current (mA)")
    plt.legend()

    # %% SECTION 3-c: Assert validity of output and return it.
    # NO MODIFICATIONS: This section should be kept as is.
    assert len(i.shape) == 1, "Currrent array should be 1-dimensional"
    return i

fitted = run_vclamp_model(os.getcwd())
```
