```python
import os 
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt(os.path.join("Data", "CookLab1UnknownChannel.txt"))
t, v, i = data[:, 0], data[:, 1], data[:, 2]
```

```python
V0 = -100  # mV: any of -100, -40, -20, 0, 20
mask = v == V0
plt.plot(t[mask] - t[mask][0], i[mask])
plt.xlabel("Time from step onset (ms)")
plt.ylabel("I")
plt.title(f"V = {V0} mV")
plt.show()
```

```python
V0 = 0  # mV: any of -100, -40, -20, 0, 20
mask = v == V0
plt.plot(t[mask] - t[mask][0], i[mask])
plt.xlabel("Time from step onset (ms)")
plt.ylabel("I")
plt.title(f"V = {V0} mV")
plt.show()
```

```python
# Voltage-step protocol: 100-sample blocks alternating holding (-60 mV) and test steps, up to t = 275 ms
n = 100
fig, (ax_v, ax_i) = plt.subplots(2, 1, sharex=True, figsize=(8, 6))
for start in range(n, 1100, 2 * n):  # start index of each test step
    sl = slice(start - n, start + 2 * n)  # holding -> step -> holding
    tt = t[sl] - t[start]
    ax_v.plot(tt, v[sl])
    ax_i.plot(tt, i[sl], label=f"{v[start]:.0f} mV")

ax_v.set_ylabel("V (mV)")
ax_i.set_ylabel("I")
ax_i.set_xlabel("Time from step onset (ms)")
ax_i.legend()
plt.show()
```
